# baselines

```python
builder=EnvBuilder()
log_dir = Path.cwd() / "logs" / "e0_random_discrete"
env=builder.make_env(space="discrete", env_idx=0, seed=0, log_dir=str(log_dir))
```

```python
n_episodes = 100
returns = []

for ep in range(n_episodes):
    obs, info = env.reset(seed=ep)
    ep_return = 0.0
    done = False

    while not done:
        action = env.action_space.sample()
        obs, reward, terminated, truncated, info = env.step(action)
        ep_return += float(reward)
        done = terminated or truncated
    returns.append(ep_return)
env.close()

returns = np.array(returns)
print(f"Mean: {returns.mean():.2f} | Std: {returns.std():.2f} | Min: {returns.min():.2f} | Max: {returns.max():.2f}")
```

```python
log_dir2 = Path.cwd() / "logs" / "e1_random_continuous"
env2=builder.make_env(space="continuous", env_idx=0, seed=0, log_dir=str(log_dir2))
```

```python
n_episodes = 100
returns = []

for ep in range(n_episodes):
    obs, info = env2.reset(seed=ep)
    ep_return = 0.0
    done = False

    while not done:
        # Samples a 2D continuous action from Box(-1.0, 1.0, (2,), float32)
        action = env2.action_space.sample()
        obs, reward, terminated, truncated, info = env2.step(action)
        ep_return += float(reward)
        done = terminated or truncated

    returns.append(ep_return)

env2.close()

returns = np.array(returns)
print(
    f"Mean: {returns.mean():.2f} | Std: {returns.std():.2f} | "
    f"Min: {returns.min():.2f} | Max: {returns.max():.2f}"
)
```
